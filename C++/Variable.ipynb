```python
class Foo:
    def __init__(self):
        self._bar=0
    @property
    def bar(self):
        return self._bar 
    @bar.setter
    def bar(self,newBar):
        self._bar=newBar
```

```python
f=Foo()
f.bar=100
print(f.bar)
```

in class based programming language, these are distingueshed into two types ; class variables (also called static member variables) 
where only one copy of the variable is shared with all instance of the class . 


instance variable : where each instances of the class has its own independent copy of the variable.
